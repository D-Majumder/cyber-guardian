```python
import json
import random
import speech_recognition as sr
import pyttsx3
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline

try:
    with open('dataset.json', 'r') as file:
        data = json.load(file)
    print("Dataset loaded successfully.")
except FileNotFoundError:
    print("Error: dataset.json not found. Make sure the file is in the same directory.")
    exit()
#Dhruba Majumder
training_sentences = []
training_labels = []
all_intents = []

for intent, content in data['intents'].items():
    all_intents.append(intent)
    for example in content['examples']:
        training_sentences.append(example)
        training_labels.append(intent)

print(f"Loaded {len(training_sentences)} training sentences.")
print(f"Identified {len(all_intents)} unique intents: {', '.join(all_intents)}")
```

```python
model = make_pipeline(TfidfVectorizer(), LogisticRegression(random_state=42, max_iter=1000))
model.fit(training_sentences, training_labels)

print("AI Model (Logistic Regression) trained successfully.")
#Dhruba Majumder
sample_input = "how to protect against viruses"
predicted_intent = model.predict([sample_input])[0]
print(f"\nSample input: '{sample_input}'")
print(f"Predicted intent: '{predicted_intent}'")
```

```python
import speech_recognition as sr
import pyttsx3

recognizer = sr.Recognizer()

def speak_response(text):
    """Initializes and runs the pyttsx3 engine for a single response."""
    try:
        engine = pyttsx3.init()
        engine.setProperty('rate', 150)
        engine.say(text)
        engine.runAndWait()
        engine.stop()
    except Exception as e:
        print(f"Error in text-to-speech: {e}")
        print(f"Fallback (text-only): {text}")
#Dhruba Majumder
def get_voice_input():
    """Listens for voice input from the user and returns the text."""
    with sr.Microphone() as source:
        print("Listening...")
        recognizer.adjust_for_ambient_noise(source)
        try:
            audio = recognizer.listen(source, timeout=5)
            user_input = recognizer.recognize_google(audio)
            print(f"You said: {user_input}")
            return user_input.lower()
        except sr.UnknownValueError:
            error_msg = "Sorry, I could not understand the audio. Please try again."
            print(error_msg)
            speak_response(error_msg)
            return None
        except sr.RequestError:
            error_msg = "Sorry, the speech service is unavailable. Please check your internet connection."
            print(error_msg)
            speak_response(error_msg)
            return None
        except sr.WaitTimeoutError:
            print("No speech detected. Timed out.")
            return None

print("CyberGuardian AI is ready! You can start speaking.")
initial_greeting = "Hello! I'm CyberGuardian AI, your assistant for digital safety. How can I help you stay safe online today?"
print(f"AI Response: {initial_greeting}")
speak_response(initial_greeting)
#Dhruba Majumder
while True:
    user_query = get_voice_input()
    if user_query:
        if "exit" in user_query or "bye" in user_query or "stop" in user_query:
            response = "Goodbye! Stay safe online."
            print(f"AI Response: {response}")
            speak_response(response)
            print("Chatbot has ended the conversation.")
            break

        predicted_intent = model.predict([user_query])[0]
        
        response_text = data['intents'][predicted_intent]['response']
        
        print(f"AI Response: {response_text}")
        speak_response(response_text)
```
